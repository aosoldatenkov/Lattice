```python
from IPython.display import display, Math

L = E_lat(8)(-1) * 2 + U_lat() * 3 + A_lat(2)(-1)
G = Genus(L)
display(Math(G.str()))

L = I_lat(1, 21)
basis = L.even_sublattice()
L = Lattice(22, L.batch_prod(basis, basis))

compl = [[[2, 0], [0, 6]], [[4, 2], [2, 4]]]
for b in compl:
    C = Lattice(2, b)
    print(C.info())
    D = DiscForm(L + C)
    iso = D.list_max_isospaces()
    for s in iso:
        M = D.overlattice(s)
        if M.parity == 0:
            print(M.info())
            G = Genus(M)
            display(Math(G.str()))
```
